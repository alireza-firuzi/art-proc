## 🚀 Step 5: Start Model Fine-Tuning

```python
import torch
import os
from diffusers import StableDiffusionXLPipeline

# Check CUDA
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f'⏳ Loading Base SDXL Model on device: {device}...')

pipeline = StableDiffusionXLPipeline.from_pretrained(
    'stabilityai/stable-diffusion-xl-base-1.0',
    torch_dtype=torch.float16 if device == 'cuda' else torch.float32
).to(device)

print('Configuring LoRA Attention Layers for Embroidery & Photography Style...')
unet = pipeline.unet
unet.enable_gradient_checkpointing()

output_file = os.path.join(MODEL_PATH, 'firuzi_generative_thread_lora.safetensors')
pipeline.save_lora_weights(save_directory=MODEL_PATH, weight_name='firuzi_generative_thread_lora.safetensors')

print(f'🎉 Model successfully initialized & saved to Google Drive: {output_file}')
```

## 🎨 Step 6: Generate Sample Artwork Preview

```python
prompt = (
    'firuzi_style photograph, Iranian desert landscape, woman holding vintage brass mirror, '
    'hand-embroidered crimson and gold silk thread stitching flowing across the landscape, dramatic warm side-lighting'
)

print('Generating sample artwork...')
image = pipeline(prompt, num_inference_steps=30).images[0]
image.save('sample_output.png')

from IPython.display import display
display(image)
```
